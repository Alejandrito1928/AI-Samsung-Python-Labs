```python
import tensorflow as tf
import tensorflow.keras as keras
from tensorflow.keras import layers


# Alternative way to define input shape using keras.Input:
# inputs = keras.Input(shape=(64, 64, 3))

model = keras.Sequential(
    [
        layers.Conv2D(32, 3, activation="relu", padding='same', input_shape=(64, 64, 3)),
        layers.MaxPooling2D(),
        layers.Conv2D(32, 3, activation="relu", padding='same'),
        layers.MaxPooling2D(),
        layers.Flatten(),
        layers.Dense(100, activation="relu", name="layer1"),
        layers.Dense(50, activation="relu", name="layer2"),
        layers.Dense(10, name="layer3"),
    ]
)
model.compile()
model.summary()
```
