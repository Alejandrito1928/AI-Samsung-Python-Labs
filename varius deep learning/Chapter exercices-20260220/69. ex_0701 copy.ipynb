```python
# Load MNIST datasets.
import tensorflow_datasets as tfds
```

```python
# Set hyperparameters: BUFFER_SIZE to shuffle data, BATCH_SIZE that is mini-batch size, and the number of training epochs.
import numpy as np
BUFFER_SIZE = 10000
BATCH_SIZE = 64
NUM_EPOCHS = 20
```

```python
# Define the number of times to repeat per epoch.
#steps_per_epoch = np.ceil(60000 / BATCH_SIZE)
# Define the number of times to repeat per epoch as an integer.
steps_per_epoch = int(np.ceil(60000 / BATCH_SIZE))
```

```python
# Define a function that preprocesses the input image and label.
def preprocess(item):
    image = item['image']
    label = item['label']
    image = tf.image.convert_image_dtype(image, tf.float32) 
    # Devolver directamente la imagen y la etiqueta (features, labels)
    return image, label[... , tf.newaxis]
```

```python
# Define two input functions for training and evaluation.
def train_input_fn():
    datasets = tfds.load(name='mnist', )
    mnist_train = datasets['train']

    dataset = mnist_train.map(preprocess)
    dataset = dataset.shuffle(BUFFER_SIZE)
    dataset = dataset.batch(BATCH_SIZE)
    return dataset.repeat()
```

```python
def eval_input_fn():
    datasets = tfds.load(name='mnist')
    mnist_test = datasets['test']
    dataset = mnist_test.map(preprocess).batch(BATCH_SIZE)
    return dataset
```

```python
# Upgraded Keras Sequential Model for Higher Accuracy
model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28, 1)), 
    
    # First Hidden Layer: Increased to 256 neurons
    tf.keras.layers.Dense(256, activation='relu'),
    # Dropout: Randomly disables 20% of neurons to prevent memorization
    tf.keras.layers.Dropout(0.2), 
    
    # Second Hidden Layer: Increased to 128 neurons
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    
    # Output Layer: 10 classes (digits 0-9)
    tf.keras.layers.Dense(10, activation='softmax')   
])

# Compile the model (same as before)
model.compile(optimizer='adam', 
              loss='sparse_categorical_crossentropy', 
              metrics=['accuracy'])

model.summary() # This will print the new structure before training
```

```python
# Obtener los datasets
train_dataset = train_input_fn()
eval_dataset = eval_input_fn()

# Entrenar el modelo llamando a Model.fit
print("Iniciando entrenamiento...")
history = model.fit(
    train_dataset,
    epochs=NUM_EPOCHS,
    steps_per_epoch=steps_per_epoch
)

# Evaluar el modelo con Model.evaluate
print("\nIniciando evaluación...")
eval_result = model.evaluate(eval_dataset, return_dict=True)
print(f"Resultado de la evaluación: {eval_result}")
```
